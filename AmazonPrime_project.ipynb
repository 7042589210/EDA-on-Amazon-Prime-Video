**HANDLING THE MISSING VALUES(CREDITS DATASETS)**-
These missing values were not actually errors but represented situations where a character name did not exist.
Instead of deleting these records or leaving them blank, we replaced the missing values with "Not Applicable".
Why Use "Not Applicable"?
Preserves all records in the dataset.
Clearly indicates that the person does not have an associated character.
Makes the dataset easier to understand and analyze.
Eliminates missing values from the column.

```python
#HANDLING MISSING VALUES
data1['character'] = data1['character'].fillna('Not Applicable')
```

```python
data1['character'].isnull().sum()
```

HANDLING THE MISSING VALUE(TITLES DATASETS)-

Step 1: Checking Missing Values
First, the number of missing values in each column was identified using:
data2.isnull().sum()
This helped determine which columns required treatment and the extent of missing data.

Step 2: Handling Categorical Columns
Description Column
data2['description'] = data2['description'].fillna('No Description Available')
Missing descriptions were replaced with "No Description Available" because description is a text field. This ensures that no records are lost and missing information is clearly indicated.
Age Certification Column
data2['age_certification'] = data2['age_certification'].fillna('Unknown')
Missing age certifications were replaced with "Unknown" because the actual rating was unavailable. This preserves the records while clearly marking missing values.
IMDb ID Column
data2['imdb_id'] = data2['imdb_id'].fillna('Not Available')
Missing IMDb IDs were replaced with "Not Available" since IMDb IDs are identifiers rather than numerical values and cannot be estimated.

Step 3: Handling the Seasons Column
data2['seasons'] = data2['seasons'].fillna(0)
The Seasons column contained a large number of missing values because movies do not have seasons. Therefore, missing values were replaced with 0, indicating that the title is most likely a movie rather than a TV show.

Step 4: Handling Numerical Columns
The following numerical columns contained missing values:
imdb_score
imdb_votes
tmdb_popularity
tmdb_score
These missing values were replaced using the median value of each column.
median_value = data2[col].median()
data2[col] = data2[col].fillna(median_value)
Why Median?
The median was chosen because:
It is less affected by extreme values (outliers).
It provides a more reliable central value than the mean for skewed data.
It preserves the overall distribution of the data.
Step 5: Verification
After handling missing values, the dataset was checked again using:
data2.isnull().sum()
The results confirmed that all missing values had been successfully treated.

```python
# Check missing values before treatment
print("Missing Values Before Handling:")
print(data2.isnull().sum())

# 1. Handle categorical columns
data2['description'] = data2['description'].fillna('No Description Available')
data2['age_certification'] = data2['age_certification'].fillna('Unknown')
data2['imdb_id'] = data2['imdb_id'].fillna('Not Available')

# 2. Handle seasons column
# Assuming missing seasons correspond mainly to Movies
data2['seasons'] = data2['seasons'].fillna(0)

# 3. Handle numerical columns using median
num_cols = ['imdb_score', 'imdb_votes', 'tmdb_popularity', 'tmdb_score']

for col in num_cols:
    median_value = data2[col].median()
    data2[col] = data2[col].fillna(median_value)

# 4. Verify missing values after treatment
print("\nMissing Values After Handling:")
print(data2.isnull().sum())
```

VISUALIZATION WITH PROBLEM STATEMENT-


```python
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (10,6)
plt.rcParams['font.size'] = 12
```

1. What type of content dominates the Amazon Prime platform?

We used a count plot(Bar chart) because the type column is categorical and we wanted to compare the number of Movies and TV Shows. The chart clearly showed that Movies are much more numerous than TV Shows, indicating that Movies dominate the Amazon Prime content library.

```python
#What type of content dominates the Amazon Prime platform?
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(6,4))
sns.countplot(x='type', data=data2 , palette='Set2')

plt.title('Distribution of Movies and TV Shows')
plt.xlabel('Content Type')
plt.ylabel('Count')
plt.show()
```

2. How has content production changed over time?
We used a line chart because release year is a time-based variable, and line charts are ideal for showing trends over time. The chart helped us observe how content production changed across years and showed that the number of titles released has generally increased, especially in recent years, indicating the growth of Amazon Prime's content catalog.

```python
#How has content production changed over time?
plt.figure(figsize=(14,6))

data2['release_year'].value_counts().sort_index().plot(
    kind='line',
    color='green',
    linewidth=3)

plt.title('Content Release Trend Over Years')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.show()
```

3. Which genres are most popular on Amazon Prime?

We used a bar chart because genres are categorical data and we wanted to compare their frequencies. After separating multiple genres using explode(), we counted the occurrences of each genre and plotted the top 10. The chart helped us identify the most popular genres available on Amazon Prime and compare their distribution easily.

```python
# Which genres are most popular on Amazon Prime?
genres = data2['genres'].explode()

plt.figure(figsize=(10,6))

genres.value_counts().head(10).plot(
    kind='bar',
    color='yellow'
)

plt.title('Top 10 Genres')
plt.xlabel('Genre')
plt.ylabel('Number of Titles')

plt.show()
```

4. What is the quality distribution of Amazon Prime content?

We used a histogram because IMDb score is a numerical variable, and histograms are ideal for showing data distribution. The chart helped us understand how ratings are spread across Amazon Prime titles and identify the score ranges where most content is concentrated. This provides insights into the overall quality of the platform's content.

```python
# What is the quality distribution of Amazon Prime content?
plt.figure(figsize=(8,5))

sns.histplot(
    data2['imdb_score'],
    bins=20,
    kde=True,
    color='red'
)

plt.title('IMDb Score Distribution')
plt.xlabel('IMDb Score')
plt.ylabel('Frequency')

plt.show()
```

5. What is the typical duration of content on Amazon Prime?

We used a histogram because runtime is a numerical variable, and histograms are useful for showing how numerical values are distributed. The chart helped us identify the most common duration range of Amazon Prime content and understand the overall spread of content lengths across the platform.


```python
# What is the typical duration of content on Amazon Prime?
plt.figure(figsize=(8,5))

sns.histplot(
    data2['runtime'],
    bins=30,
    color='mediumpurple'
)

plt.title('Runtime Distribution')
plt.xlabel('Runtime (Minutes)')
plt.ylabel('Count')

plt.show()
```

6. Which audience groups are targeted the most?

We used a bar chart because age certification is a categorical variable, and bar charts are ideal for comparing category frequencies. The chart helped us identify which audience group receives the most content on Amazon Prime by showing the number of titles under each age certification category.

```python
# Which audience groups are targeted the most?
plt.figure(figsize=(10,5))
colors = [
    'olive',
    'orange',
    'green',
    'yellow',
    'brown',
    'red',
    'pink',
    'gray',
    'skyblue',
    'purple'
]

data2['age_certification'].value_counts().plot(kind='bar',color=colors)

plt.title('Age Certification Distribution')
plt.xlabel('Certification')
plt.ylabel('Count')

plt.show()
```

7. Which countries contribute the most content?

We used a bar chart because production countries are categorical data and we wanted to compare their content contributions. After separating multiple countries using explode(), we counted the number of titles from each country and plotted the top 10. The chart helped us identify the countries that contribute the most content to Amazon Prime.

```python
# Which countries contribute the most content?
countries = data2['production_countries'].explode()

plt.figure(figsize=(12,6))

colors = [
    'purple',
    'green',
    'red',
    'brown',
    'orange',
    'blue',
    'pink',
    'gray',
    'skyblue',
    'seagreen'
]

countries.value_counts().head(10).plot(
    kind='bar',
    color=colors,
    edgecolor='black'
)

plt.title('Top 10 Production Countries')
plt.xlabel('Country')
plt.ylabel('Number of Titles')

plt.show()
```
